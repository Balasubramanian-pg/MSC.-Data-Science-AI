## SECTION 3: First Bokeh Plot - The Foundation

```python
# ============================================================================
# SECTION 3: YOUR FIRST BOKEH PLOT
# ============================================================================

"""
CREATING A BASIC PLOT:
• figure() = Creates the canvas (like plt.figure())
• p.line() = Adds a line glyph (like plt.plot())
• p.circle() = Adds scatter points (like plt.scatter())
• show() = Renders the plot (like plt.show())
"""

# Generate sample data
x_data = np.linspace(0, 10, 100)
y_data = np.sin(x_data) + np.random.normal(0, 0.1, 100)

# Create figure
p = figure(
    title="My First Bokeh Plot: Sine Wave with Noise",
    x_axis_label="X-axis (Time)",
    y_axis_label="Y-axis (Value)",
    width=800,
    height=400,
    tools="pan,box_zoom,wheel_zoom,reset,save",
    toolbar_location="above"
)

# Add line
p.line(
    x_data, 
    y_data, 
    legend_label="Sine Wave", 
    line_width=3, 
    line_color="navy"
)

# Add scatter points
p.scatter(
    x_data, 
    y_data, 
    legend_label="Data Points", 
    size=6, 
    color="orange", 
    alpha=0.6
)

# Style the plot
p.title.text_font_size = "16pt"
p.title.text_font_style = "bold"
p.xaxis.axis_label_text_font_size = "14pt"
p.yaxis.axis_label_text_font_size = "14pt"
p.legend.location = "top_right"
p.legend.background_fill_alpha = 0.8
p.grid.grid_line_alpha = 0.3

show(p)

print("""
PLOT FEATURES:
────────────
• Zoom: Use mouse wheel or box zoom
• Pan: Click and drag
• Reset: Restore original view
• Save: Download as PNG
• Hover: Shows point values (not yet configured)
""")
```

## SECTION 4: Understanding Glyphs - The Building Blocks

```python
# ============================================================================
# SECTION 4: GLYPHS - THE VISUAL BUILDING BLOCKS
# ============================================================================

"""
GLYPHS EXPLAINED:
Glyphs are the visual objects rendered on the canvas.

COMMON GLYPH TYPES:
1. line()      - Continuous lines
2. circle()    - Circular markers
3. square()    - Square markers
4. triangle()  - Triangular markers
5. vbar()      - Vertical bars
6. hbar()      - Horizontal bars
7. patch()     - Filled polygons
8. rect()      - Rectangles
9. text()      - Text annotations
10. image()    - Image display
"""

# Create different datasets
x = np.linspace(0, 10, 20)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

# Create figure
p = figure(
    title="Multiple Glyphs Demonstration",
    width=800,
    height=500,
    tools="pan,wheel_zoom,reset,hover",
    toolbar_location="above"
)

# 1. Line glyph
p.line(x, y1, legend_label="Line Glyph", line_width=3, color="navy")

# 2. Circle glyph
p.circle(x, y1, legend_label="Circle Glyph", size=10, color="orange", alpha=0.7)

# 3. Square glyph
p.square(x, y2, legend_label="Square Glyph", size=12, color="green", alpha=0.6)

# 4. Triangle glyph
p.triangle(x, y3, legend_label="Triangle Glyph", size=14, color="red", alpha=0.5)

# 5. Patch glyph (filled area)
p.patch(x, y3, legend_label="Patch Glyph", color="red", alpha=0.2)

# 6. Vertical bar glyph
from bokeh.transform import dodge
categories = ['A', 'B', 'C', 'D']
values = [5, 3, 8, 6]
p.vbar(
    x=categories, 
    top=values, 
    width=0.5, 
    legend_label="Bar Glyph",
    color="purple",
    alpha=0.7
)

p.legend.location = "top_left"
p.legend.click_policy = "hide"

show(p)

print("""
GLYPH TYPES REFERENCE:
────────────────────
• Line: Continuous data representation
• Circle: Point data with size variation
• Square: Alternative point marker
• Triangle: Alternative point marker
• Patch: Filled areas (useful for confidence bands)
• Bar: Categorical data representation
• Rectangle: Grid-based visualizations
• Text: Annotations and labels
• Image: Raster data display
""")
```

## SECTION 5: ColumnDataSource - The Data Engine

```python
# ============================================================================
# SECTION 5: COLUMNDATASOURCE - THE DATA ENGINE
# ============================================================================

"""
COLUMNDATASOURCE EXPLAINED:

What is ColumnDataSource?
• A data structure that maps column names to sequences of values
• The bridge between Python data and Bokeh's JavaScript rendering
• Enables dynamic updates and data sharing between glyphs

WHY USE IT:
• More efficient for large datasets
• Enables data streaming
• Supports linked brushing
• Allows data-driven styling
• Makes interactive updates possible
"""

# Create data
data = {
    'x': np.linspace(0, 10, 100),
    'y': np.sin(np.linspace(0, 10, 100)),
    'y2': np.cos(np.linspace(0, 10, 100)),
    'color': ['#%02x%02x%02x' % (int(r*255), int(g*255), int(b*255)) 
              for r, g, b in zip(np.random.rand(100), 
                                 np.random.rand(100), 
                                 np.random.rand(100))],
    'size': np.random.randint(5, 20, 100),
    'label': [f'Point {i}' for i in range(100)]
}

# Create ColumnDataSource
source = ColumnDataSource(data=data)

# Create figure
p = figure(
    title="ColumnDataSource - Data-Driven Visualization",
    width=800,
    height=500,
    tools="pan,wheel_zoom,reset,box_select,lasso_select,hover",
    toolbar_location="above"
)

# Add line
p.line('x', 'y', source=source, legend_label="Line", line_width=3, color="black")

# Add scatter with data-driven properties
p.scatter(
    'x', 'y2', 
    source=source,
    legend_label="Scatter",
    size='size',  # Size from data
    color='color',  # Color from data
    alpha=0.7
)

# Add hover tool
hover = HoverTool(
    tooltips=[
        ("Index", "$index"),
        ("X", "@x{0.00}"),
        ("Y", "@y{0.00}"),
        ("Y2", "@y2{0.00}"),
        ("Label", "@label"),
        ("Size", "@size"),
        ("Color", "@color")
    ]
)
p.add_tools(hover)

p.legend.location = "top_left"

show(p)

print("""
COLUMNDATASOURCE ADVANTAGES:
──────────────────────────
1. Single source for multiple glyphs
2. Data-driven visual properties (size, color, alpha)
3. Efficient for large datasets
4. Enables streaming updates
5. Supports linked selections
6. JSON serialization for web delivery

BEST PRACTICES:
• Always use ColumnDataSource for interactive plots
• Name columns descriptively (x, y, size, color, etc.)
• Use @column_name in tooltips
• Update data via source.data = new_data
""")
```

## SECTION 6: Interactive Tools - Zoom, Pan, Hover

```python
# ============================================================================
# SECTION 6: INTERACTIVE TOOLS - ZOOM, PAN, HOVER
# ============================================================================

"""
INTERACTIVE TOOLS OVERVIEW:

AVAILABLE TOOLS:
• PanTool: Click and drag to pan
• WheelZoomTool: Scroll to zoom
• BoxZoomTool: Drag box to zoom
• ResetTool: Reset view to original
• SaveTool: Download as PNG
• HoverTool: Show data on hover
• TapTool: Select on click
• BoxSelectTool: Select by dragging box
• LassoSelectTool: Select by drawing freeform
• CrosshairTool: Crosshair cursor
• UndoTool: Undo last action
• RedoTool: Redo last action
"""

# Generate data
x = np.random.normal(0, 1, 200)
y = np.random.normal(0, 1, 200)
colors = np.random.choice(['red', 'blue', 'green'], 200)
sizes = np.random.randint(8, 25, 200)

source = ColumnDataSource(data={
    'x': x,
    'y': y,
    'color': colors,
    'size': sizes,
    'label': [f'Point {i}' for i in range(200)]
})

# Create figure with multiple tools
p = figure(
    title="Multiple Interactive Tools Demo",
    width=800,
    height=500,
    tools="pan,box_zoom,wheel_zoom,reset,save,box_select,lasso_select,tap,crosshair,hover",
    toolbar_location="above"
)

# Add scatter plot
p.scatter(
    'x', 'y',
    source=source,
    size='size',
    color='color',
    alpha=0.7,
    legend_label="Data Points"
)

# Configure hover tool
hover = HoverTool(
    tooltips=[
        ("Index", "$index"),
        ("X", "@x{0.00}"),
        ("Y", "@y{0.00}"),
        ("Color", "@color"),
        ("Size", "@size"),
        ("Label", "@label")
    ],
    mode='mouse',
    point_policy='follow_mouse'
)
p.add_tools(hover)

# Style
p.xaxis.axis_label = "X Dimension"
p.yaxis.axis_label = "Y Dimension"
p.grid.grid_line_alpha = 0.3

show(p)

print("""
TOOL REFERENCE GUIDE:
───────────────────
TOOL                  FUNCTION
─────────────────────────────────────────────
Pan                  Click and drag to pan
Wheel Zoom           Scroll to zoom in/out
Box Zoom             Drag box to zoom into area
Reset                Reset to original view
Save                 Save as PNG image
Box Select           Select points by dragging box
Lasso Select         Select points by drawing lasso
Tap                  Select point on click
Hover                Show tooltips on hover
Crosshair            Crosshair cursor for precision
Undo/Redo            Undo/redo zoom/pan actions
""")
```

## SECTION 7: Hover Tool Configuration - Deep Dive

```python
# ============================================================================
# SECTION 7: HOVER TOOL - DEEP DIVE CONFIGURATION
# ============================================================================

"""
HOVER TOOL CONFIGURATION:

TOOLTIP SYNTAX:
• @{column}: Display column value
• $index: Display index
• $name: Display glyph name
• {format}: Format numeric values

FORMATTING OPTIONS:
• {0.00}: 2 decimal places
• {0.000}: 3 decimal places
• {int}: Integer
• {0.0%}: Percentage
• {0.0a}: SI prefix
• {datetime}: Date formatting

HOVER MODES:
• 'mouse': Tooltip follows mouse cursor
• 'vline': Vertical line at mouse position
• 'hline': Horizontal line at mouse position
"""

# Create complex dataset
n = 100
x = np.linspace(0, 100, n)
y = np.sin(x/10) * np.exp(-x/50) + np.random.normal(0, 0.05, n)

source = ColumnDataSource(data={
    'x': x,
    'y': y,
    'category': np.random.choice(['A', 'B', 'C', 'D'], n),
    'value': np.random.randint(1, 100, n),
    'percentage': np.random.rand(n) * 100,
    'date': pd.date_range('2023-01-01', periods=n, freq='D').strftime('%Y-%m-%d'),
    'large_number': np.random.randint(1000, 1000000, n)
})

p = figure(
    title="Advanced Hover Tool Configuration",
    width=900,
    height=500,
    tools="pan,wheel_zoom,reset,hover",
    toolbar_location="above"
)

# Add line
p.line('x', 'y', source=source, line_width=3, color="navy", legend_label="Signal")

# Add scatter with enhanced hover
p.circle(
    'x', 'y',
    source=source,
    size=8,
    color='category',
    alpha=0.6,
    legend_label="Data Points"
)

# Advanced hover configuration
hover = HoverTool(
    tooltips="""
    <div style="background-color: white; padding: 10px; border: 1px solid #ccc;">
        <h4>Point Information</h4>
        <hr>
        <span style="font-weight: bold;">Index:</span> $index<br>
        <span style="font-weight: bold;">Date:</span> @date<br>
        <span style="font-weight: bold;">X:</span> @x{0.00}<br>
        <span style="font-weight: bold;">Y:</span> @y{0.000}<br>
        <span style="font-weight: bold;">Category:</span> @category<br>
        <span style="font-weight: bold;">Value:</span> @value{int}<br>
        <span style="font-weight: bold;">Percentage:</span> @percentage{0.0%}<br>
        <span style="font-weight: bold;">Large Number:</span> @large_number{0.0a}
    </div>
    """,
    mode='mouse',
    point_policy='follow_mouse',
    renderers=[p.renderers[-1]]  # Apply only to scatter points
)
p.add_tools(hover)

p.legend.location = "top_left"
p.legend.click_policy = "hide"

show(p)

print("""
HOVER TOOL BEST PRACTICES:
─────────────────────────
1. Use HTML formatting for styled tooltips
2. Display relevant data only
3. Format numbers appropriately (decimals, percentages)
4. Use mode='mouse' for point data
5. Use mode='vline' for time series data
6. Apply to specific renderers when needed
7. Keep tooltips concise but informative
""")
```

## SECTION 9: Color Palettes and Theming

```python
# ============================================================================
# SECTION 9: COLOR PALETTES AND THEMING
# ============================================================================

"""
BOKEH COLOR PALETTES:

CATEGORICAL PALETTES:
• Category10: 10 distinct colors (colorblind friendly)
• Category20: 20 distinct colors
• Category20b: 20 distinct colors (alternative)
• Category20c: 20 distinct colors (alternative)

SEQUENTIAL PALETTES:
• Blues, Greens, Reds, Purples, Oranges
• Viridis, Plasma, Inferno, Magma, Cividis
• Gray: 256 grayscale values

DIVERGING PALETTES:
• RdBu: Red-Blue
• RdGy: Red-Gray
• Spectral: Full spectrum

CUSTOM PALETTES:
• hex color codes
• rgb/rgba tuples
• Named colors
"""

from bokeh.palettes import Category10, Category20, Viridis256, Plasma256, Spectral6

# Generate data
categories = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']
values1 = np.random.randint(10, 50, 8)
values2 = np.random.randint(10, 50, 8)
values3 = np.random.randint(10, 50, 8)

# Create a figure
p = figure(
    title="Color Palette Demonstration",
    width=900,
    height=500,
    x_range=categories,
    tools="pan,wheel_zoom,reset,hover",
    toolbar_location="above"
)

# 1. Category10 palette (default)
p.vbar(
    x=dodge(categories, -0.25, range=p.x_range),
    top=values1,
    width=0.2,
    color=Category10[8],
    legend_label="Category10",
    alpha=0.8
)

# 2. Category20 palette
p.vbar(
    x=dodge(categories, 0.0, range=p.x_range),
    top=values2,
    width=0.2,
    color=Category20[8],
    legend_label="Category20",
    alpha=0.8
)

# 3. Custom palette from Viridis
viridis_colors = [Viridis256[int(i * 256/8)] for i in range(8)]
p.vbar(
    x=dodge(categories, 0.25, range=p.x_range),
    top=values3,
    width=0.2,
    color=viridis_colors,
    legend_label="Viridis",
    alpha=0.8
)

p.legend.location = "top_right"
p.xaxis.axis_label = "Categories"
p.yaxis.axis_label = "Values"

show(p)

# 2nd plot: Sequential palette demonstration
# Create scatter plot with sequential coloring
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)  # Color value

# Map values to color indices
colors = [Viridis256[int(val * 255)] for val in z]

p2 = figure(
    title="Sequential Color Mapping (Viridis)",
    width=800,
    height=400,
    tools="pan,wheel_zoom,reset,hover",
    toolbar_location="above"
)

p2.scatter(
    x, y,
    size=z * 30 + 10,
    color=colors,
    alpha=0.7,
    legend_label="Data Points"
)

p2.xaxis.axis_label = "X"
p2.yaxis.axis_label = "Y"

show(p2)

print("""
COLOR PALETTE SELECTION GUIDE:
─────────────────────────────
PURPOSE              RECOMMENDED PALETTES
─────────────────────────────────────────
Categorical Data     Category10, Category20
Sequential Data      Viridis, Plasma, Blues, Reds
Diverging Data       RdBu, RdGy, Spectral
Colorblind Safe      Category10, Viridis
Monochromatic        Gray, Blues, Greens
Custom Branding      User-defined hex codes

THEMING IN BOKEH:
• Built-in themes: 'caliber', 'dark_minimal', 'light_minimal'
• Custom themes: JSON configuration
• Consistent styling across plots
• Easy theme switching
""")
```

## SECTION 14: CustomJS for Interactive Updates

```python
# ============================================================================
# SECTION 14: CUSTOMJS FOR INTERACTIVE UPDATES
# ============================================================================

"""
CUSTOMJS EXPLAINED:

WHAT IS CUSTOMJS?
• JavaScript code that runs in the browser
• Connects Python widgets to plot updates
• Enables real-time interactivity without server trips
• Uses the BokehJS API

CUSTOMJS PATTERNS:
1. Update data: Modify ColumnDataSource contents
2. Update properties: Change visual attributes
3. Callbacks: React to widget changes
4. Animation: Create dynamic visualizations
"""

from bokeh.models import CustomJS, Slider, Button

# Create data
x = np.linspace(0, 10, 100)
y = np.sin(x)

source = ColumnDataSource(data={'x': x, 'y': y})

# Create plot
p = figure(
    title="CustomJS Interactive Updates",
    width=800,
    height=400,
    tools="pan,wheel_zoom,reset"
)

p.line('x', 'y', source=source, line_width=3, color="#3498db")

# Slider for frequency
slider = Slider(start=1, end=10, step=0.1, value=1, title="Frequency", width=300)

# Slider for amplitude
amp_slider = Slider(start=0.1, end=2, step=0.1, value=1, title="Amplitude", width=300)

# Button for reset
reset_button = Button(label="Reset to Sine", button_type="warning")

# Advanced CustomJS with multiple callbacks
callback = CustomJS(
    args=dict(
        source=source,
        x=x,
        freq_slider=slider,
        amp_slider=amp_slider
    ),
    code="""
        const freq = freq_slider.value;
        const amp = amp_slider.value;
        const data = source.data;
        const y = data['y'];
        
        for (let i = 0; i < x.length; i++) {
            y[i] = amp * Math.sin(freq * x[i]);
        }
        source.change.emit();
    """
)

# Reset callback
reset_callback = CustomJS(
    args=dict(
        source=source,
        x=x,
        freq_slider=slider,
        amp_slider=amp_slider
    ),
    code="""
        const data = source.data;
        const y = data['y'];
        
        for (let i = 0; i < x.length; i++) {
            y[i] = Math.sin(x[i]);
        }
        freq_slider.value = 1;
        amp_slider.value = 1;
        source.change.emit();
    """
)

# Connect callbacks
slider.js_on_change('value', callback)
amp_slider.js_on_change('value', callback)
reset_button.js_on_click(reset_callback)

# Layout
layout = column(
    p,
    row(slider, amp_slider),
    reset_button,
    sizing_mode='stretch_width'
)

show(layout)

print("""
CUSTOMJS BEST PRACTICES:
─────────────────────────
1. Keep JavaScript code clean and efficient
2. Use console.log() for debugging
3. Pass all needed objects through args
4. Use source.change.emit() to trigger updates
5. Handle edge cases (null values, divisions by zero)
6. Test interactions thoroughly
7. Document complex callback logic

COMMON PATTERNS:
• Data transformation: modify array values
• Property updates: change colors, sizes
• Animation: update data in loops with setTimeout
• Filtering: show/hide data points
""")
```

## SECTION 16: Animated Visualizations

```python
# ============================================================================
# SECTION 16: ANIMATED VISUALIZATIONS
# ============================================================================

"""
ANIMATION TECHNIQUES:

1. Periodic Updates: Scheduled data changes
2. Slider Animation: Controlled by user
3. Time Series Animation: Moving window over data
4. Wave Animation: Propagating signals
"""

# Create data
x = np.linspace(0, 10, 200)
y = np.sin(x)

source = ColumnDataSource(data={'x': x, 'y': y})

# Create plot
p = figure(
    title="Animated Wave Visualization",
    width=800,
    height=400,
    tools="pan,wheel_zoom,reset"
)

p.line('x', 'y', source=source, line_width=3, color="#3498db")

# Animation slider
anim_slider = Slider(start=0, end=2*np.pi, step=0.01, value=0, title="Phase Shift", width=300)

# Speed control
speed_slider = Slider(start=0.1, end=2, step=0.1, value=0.5, title="Animation Speed", width=300)

# Animation callback
anim_callback = CustomJS(
    args=dict(source=source, x=x, phase_slider=anim_slider, speed_slider=speed_slider),
    code="""
        const phase = phase_slider.value;
        const data = source.data;
        const y = data['y'];
        
        for (let i = 0; i < x.length; i++) {
            y[i] = Math.sin(x[i] + phase);
        }
        source.change.emit();
    """
)

anim_slider.js_on_change('value', anim_callback)

# Animation button
play_button = Button(label="▶ Play", button_type="success")

# Auto-animation callback (JavaScript-based)
play_callback = CustomJS(
    args=dict(phase_slider=anim_slider, speed_slider=speed_slider),
    code="""
        let running = false;
        let interval_id = null;
        
        function animate() {
            const speed = speed_slider.value;
            let phase = phase_slider.value + 0.02 * speed;
            if (phase > 2 * Math.PI) {
                phase = 0;
            }
            phase_slider.value = phase;
            phase_slider.change.emit();
        }
        
        if (cb_obj.label === '▶ Play') {
            cb_obj.label = '⏸ Pause';
            cb_obj.button_type = 'warning';
            interval_id = setInterval(animate, 50);
        } else {
            cb_obj.label = '▶ Play';
            cb_obj.button_type = 'success';
            clearInterval(interval_id);
        }
    """
)

play_button.js_on_click(play_callback)

# Layout
layout = column(
    p,
    row(anim_slider, speed_slider),
    play_button,
    sizing_mode='stretch_width'
)

show(layout)

print("""
ANIMATION TECHNIQUES:
─────────────────────
1. Phase Animation: Moving waves
2. Time-Lapse: Data evolution over time
3. Particle Animation: Moving points
4. Chart Animation: Growing datasets
5. Transition Animation: Smooth changes

BEST PRACTICES:
• Use requestAnimationFrame for smooth animations
• Limit frame rate for performance
• Provide play/pause controls
• Include speed controls
• Handle cleanup when animation stops
""")
```

```python
# ============================================================================
# SECTION 23: INTERACTIVE DASHBOARD WITH MULTIPLE COMPONENTS
# ============================================================================

"""
DASHBOARD COMPONENTS:

1. Plots: Multiple linked visualizations
2. Widgets: Controls for interaction
3. Layout: Organized component arrangement
4. Data: Shared data source
5. Interactivity: Cross-component updates
"""

# Generate sample data
np.random.seed(42)
n = 200

dashboard_data = {
    'x': np.random.rand(n) * 10,
    'y': np.random.rand(n) * 10,
    'z': np.random.rand(n) * 10,
    'category': np.random.choice(['Product A', 'Product B', 'Product C'], n),
    'sales': np.random.randint(100, 1000, n),
    'profit': np.random.randint(10, 200, n)
}

source = ColumnDataSource(data=dashboard_data)

# 1. Scatter plot
p_scatter = figure(
    title="Sales vs Profit",
    width=400,
    height=350,
    tools="pan,wheel_zoom,reset,hover,box_select,lasso_select",
    toolbar_location="above"
)

p_scatter.scatter(
    'sales', 'profit',
    source=source,
    size=8,
    color='category',
    alpha=0.6,
    legend_label='category'
)

p_scatter.xaxis.axis_label = "Sales"
p_scatter.yaxis.axis_label = "Profit"
p_scatter.legend.click_policy = "hide"

# 2. Bar chart
p_bar = figure(
    title="Sales by Category",
    width=400,
    height=350,
    x_range=['Product A', 'Product B', 'Product C'],
    tools="pan,wheel_zoom,reset",
    toolbar_location="above"
)

# Aggregate data for bar chart
agg_data = pd.DataFrame(dashboard_data).groupby('category')['sales'].sum().reset_index()
source_bar = ColumnDataSource(agg_data)

p_bar.vbar(
    x='category',
    top='sales',
    source=source_bar,
    width=0.6,
    color=['#3498db', '#e74c3c', '#2ecc71'],
    alpha=0.8
)

p_bar.xaxis.axis_label = "Category"
p_bar.yaxis.axis_label = "Total Sales"

# 3. Histogram
p_hist = figure(
    title="Profit Distribution",
    width=400,
    height=350,
    tools="pan,wheel_zoom,reset",
    toolbar_location="above"
)

profit_hist, profit_edges = np.histogram(dashboard_data['profit'], bins=20)
source_hist = ColumnDataSource(data={
    'top': profit_hist,
    'bottom': 0,
    'left': profit_edges[:-1],
    'right': profit_edges[1:]
})

p_hist.quad(
    top='top', bottom='bottom', left='left', right='right',
    source=source_hist,
    color="#9b59b6",
    alpha=0.7
)

p_hist.xaxis.axis_label = "Profit"
p_hist.yaxis.axis_label = "Frequency"

# 4. Widgets
category_select = Select(options=['All', 'Product A', 'Product B', 'Product C'], value='All', title="Filter by Category:", width=200)

sales_slider = RangeSlider(start=0, end=1000, value=(100, 900), step=10, title="Sales Range:", width=200)

reset_button = Button(label="Reset Filters", button_type="warning", width=200)

# Widget callbacks
callback_filter = CustomJS(
    args=dict(
        source=source,
        category_select=category_select,
        sales_slider=sales_slider,
        original_data=dashboard_data
    ),
    code="""
        const category = category_select.value;
        const [min_sales, max_sales] = sales_slider.value;
        const data = original_data;
        
        const new_data = {
            'x': [], 'y': [], 'z': [],
            'category': [], 'sales': [], 'profit': []
        };
        
        for (let i = 0; i < data['x'].length; i++) {
            const matches_category = category === 'All' || data['category'][i] === category;
            const matches_sales = data['sales'][i] >= min_sales && data['sales'][i] <= max_sales;
            
            if (matches_category && matches_sales) {
                for (const key in data) {
                    new_data[key].push(data[key][i]);
                }
            }
        }
        
        source.data = new_data;
        source.change.emit();
    """
)

category_select.js_on_change('value', callback_filter)
sales_slider.js_on_change('value', callback_filter)

reset_callback = CustomJS(
    args=dict(
        source=source,
        original_data=dashboard_data,
        category_select=category_select,
        sales_slider=sales_slider
    ),
    code="""
        source.data = original_data;
        category_select.value = 'All';
        sales_slider.value = [0, 1000];
        source.change.emit();
    """
)

reset_button.js_on_click(reset_callback)

# Layout dashboard
widget_panel = column(
    Div(text="<h3>Dashboard Controls</h3>"),
    category_select,
    sales_slider,
    reset_button,
    sizing_mode='stretch_width'
)

plot_grid = gridplot(
    [[p_scatter, p_bar],
     [p_hist, widget_panel]],
    toolbar_location='above',
    merge_tools=True
)

show(plot_grid)

print("""
DASHBOARD BEST PRACTICES:
─────────────────────────
1. Keep layout clean and organized
2. Use consistent color schemes
3. Provide clear labels and titles
4. Use interactive filters for exploration
5. Link multiple views for coordinated analysis
6. Include reset functionality
7. Use responsive sizing

DASHBOARD COMPONENTS:
• Plots: Primary visualizations
• Widgets: Controls and filters
• Text: Titles and descriptions
• Data: Shared data sources
• Layout: Organized presentation
""")
```

## SECTION 25: Performance Optimization

```python
# ============================================================================
# SECTION 25: PERFORMANCE OPTIMIZATION
# ============================================================================

"""
PERFORMANCE OPTIMIZATION TECHNIQUES:

1. Data Sampling: Reduce point count
2. WebGL Acceleration: Hardware rendering
3. Data Aggregation: Pre-aggregate large datasets
4. Lazy Loading: Load data on demand
5. Efficient Data Structures: Use ColumnDataSource
6. Datashader Integration: Large-scale visualization
"""

# Compare performance: Full dataset vs sampled

# Generate large dataset
n_large = 50000
x_large = np.random.rand(n_large)
y_large = np.random.rand(n_large)

# 1. Full dataset (may be slow)
p1 = figure(
    title=f"Full Dataset ({n_large} points)",
    width=400,
    height=300,
    tools="pan,wheel_zoom,reset"
)

p1.scatter(x_large, y_large, size=2, alpha=0.5, color="#3498db")

# 2. Sampled dataset (10%)
sample_idx = np.random.choice(n_large, int(n_large * 0.1), replace=False)
x_sample = x_large[sample_idx]
y_sample = y_large[sample_idx]

p2 = figure(
    title=f"Sampled Dataset ({len(sample_idx)} points)",
    width=400,
    height=300,
    tools="pan,wheel_zoom,reset"
)

p2.scatter(x_sample, y_sample, size=3, alpha=0.7, color="#e74c3c")

# 3. Aggregated data (hexbin)
from bokeh.models import HBar

# Create 2D histogram
hist2d, xedges, yedges = np.histogram2d(x_large, y_large, bins=30)
x_centers = (xedges[:-1] + xedges[1:]) / 2
y_centers = (yedges[:-1] + yedges[1:]) / 2

# Prepare aggregated data
x_flat = []
y_flat = []
counts_flat = []

for i in range(len(x_centers)):
    for j in range(len(y_centers)):
        if hist2d[i, j] > 0:
            x_flat.append(x_centers[i])
            y_flat.append(y_centers[j])
            counts_flat.append(hist2d[i, j])

p3 = figure(
    title=f"Aggregated ({len(counts_flat)} bins)",
    width=400,
    height=300,
    tools="pan,wheel_zoom,reset"
)

# Color mapping for counts
max_count = max(counts_flat)
colors = [Viridis256[int((c/max_count) * 255)] for c in counts_flat]

p3.scatter(
    x_flat, y_flat,
    size=[c/max_count * 20 + 2 for c in counts_flat],
    color=colors,
    alpha=0.8
)

# Display comparison
grid = gridplot(
    [[p1, p2, p3]],
    toolbar_location='above',
    merge_tools=True
)

show(grid)

print("""
PERFORMANCE OPTIMIZATION STRATEGIES:
────────────────────────────────────
1. DATA SAMPLING: Display representative subset
2. DATA AGGREGATION: Pre-aggregate into bins
3. WEBGL: Enable for large scatter plots
4. DATASHADER: For billion-point visualizations
5. STREAMING: Use rollover limits
6. PROGRESSIVE RENDERING: Show data incrementally

RECOMMENDATIONS:
• < 1,000 points: No optimization needed
• 1,000 - 10,000: Use WebGL
• 10,000 - 100,000: Sample or aggregate
• > 100,000: Use Datashader

OPTIMIZATION CHECKLIST:
□ Use ColumnDataSource
□ Enable WebGL (output_backend='webgl')
□ Sample large datasets
□ Aggregate when possible
□ Limit tooltips on large datasets
□ Use responsive sizing
""")
```
